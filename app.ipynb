```python
%%writefile emoji_translator.py
import emoji

def translate_user_input(raw_text):
    text_with_tags = emoji.demojize(raw_text)
    clean_text = text_with_tags.replace(":", " ").replace("_", " ")
    return " ".join(clean_text.split()).lower()
```

```python
%%writefile empathy_matrix.py
import random

def get_empathetic_response(emotion):
    responses = {
        "Joy": ["That is incredible! Tell me more! 🎉", "Your energy is contagious!"],
        "Sadness": ["I'm so sorry. Take a deep breath, I'm here to listen. 💙", "You don't have to go through it alone."],
        "Neutral": ["I'm here and listening. How is your day going?", "Got it. What's on your mind?"]
    }
    if emotion not in responses:
        emotion = "Neutral"
    return random.choice(responses[emotion])
```

```python
%%writefile app.py
import streamlit as st
from empathy_matrix import get_empathetic_response
from emoji_translator import translate_user_input

st.set_page_config(page_title="Echo-Emote", page_icon="🧠")
st.title("🧠 Echo-Emote AI")

if "messages" not in st.session_state:
    st.session_state.messages = [{"role": "assistant", "content": "Hi there! I'm listening. How are you feeling today?"}]

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])

user_input = st.chat_input("Type your message here (with emojis!)...")

if user_input:
    with st.chat_message("user"):
        st.markdown(user_input)
    st.session_state.messages.append({"role": "user", "content": user_input})

    clean_text = translate_user_input(user_input)

    if "crying" in clean_text or "sad" in clean_text:
        detected_emotion = "Sadness"
    elif "joy" in clean_text or "happy" in clean_text:
        detected_emotion = "Joy"
    else:
        detected_emotion = "Neutral"

    ai_response = get_empathetic_response(detected_emotion)

    with st.chat_message("assistant"):
        st.markdown(ai_response)
    st.session_state.messages.append({"role": "assistant", "content": ai_response})
```
