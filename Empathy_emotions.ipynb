```python
import random

def get_empathetic_response(emotion):
    """
    The Deep Intel Empathy Matrix.
    Routes the AI's detected emotion to a context-aware, empathetic response.
    """
    responses = {
        "Joy": [
            "That is incredible! I can hear the excitement! Tell me more! 🎉",
            "I love hearing that! You absolutely deserve to celebrate this. 😊",
            "Your energy is contagious! What's making you feel so great?"
        ],
        "Sadness": [
            "I'm so sorry you're feeling this way. Take a deep breath, I'm here to listen. 💙",
            "It sounds like you're going through a really tough time. You don't have to go through it alone.",
            "That sounds really heavy. Take all the time you need; it's okay to not be okay right now."
        ],
        "Anger": [
            "I can completely understand why you're frustrated. That sounds incredibly annoying.",
            "It is totally valid to be upset about this. Take a breath, do you want to vent about it?",
            "That sounds really unfair. I'm listening if you want to let it all out."
        ],
        "Fear": [
            "It is completely natural to feel anxious about this. I'm right here with you.",
            "That sounds really overwhelming. Let's take this one step at a time.",
            "You are safe. It's okay to feel scared, but you don't have to face it alone."
        ],
        "Disgust": [
            "Ugh, I totally get why that would make you feel that way. That's awful.",
            "I completely agree, that sounds incredibly unpleasant.",
            "You have every right to be put off by that."
        ],
        "Surprise": [
            "Wow, I didn't see that coming either! How are you processing that? 😲",
            "That is wildly unexpected! Are you doing okay after hearing that?",
            "Oh my goodness! What a curveball. Tell me everything!"
        ],
        "Neutral": [
            "I'm here and listening. How is the rest of your day going?",
            "Got it. Is there anything specific on your mind you want to chat about?",
            "I understand. What's been the highlight of your week so far?"
        ]
    }

    # If for some reason an unknown emotion slips through, default to neutral
    if emotion not in responses:
        emotion = "Neutral"

    return random.choice(responses[emotion])

# Quick test to make sure the matrix works if we run this file directly!
if __name__ == "__main__":
    print("Testing the Deep Intel Empathy Matrix...")
    print(f"Test (Joy): {get_empathetic_response('Joy')}")
    print(f"Test (Sadness): {get_empathetic_response('Sadness')}")

    # ... (Keep the dictionary and get_empathetic_response function exactly the same above this)

# The Interactive Terminal Test
if __name__ == "__main__":
    print("🧠 Deep Intel Empathy Matrix - Terminal Test Booting Up...")
    print("Type an emotion (Joy, Sadness, Anger, Fear, Disgust, Surprise, Neutral) to see the AI's response.")
    print("Type 'exit' to shut down the matrix.\n")

    while True:
        # We take your input and capitalize it so it matches the dictionary exactly
        detected_emotion = input("🧪 Simulate AI Detection (Enter emotion): ").strip().capitalize()

        if detected_emotion.lower() == 'exit':
            print("Shutting down the matrix. Good work, Lead Engineer.")
            break

        # Route the input through your custom Deep Intel logic
        ai_reply = get_empathetic_response(detected_emotion)
        print(f"🤖 Echo-Emote AI: {ai_reply}\n")
```
